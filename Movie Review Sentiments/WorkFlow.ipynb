```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv('IMDB Dataset.csv')
```

```python
data.head()
```

```python
data['review'][0]
```

```python
import re
```

```python
pattern = re.compile(r'<.*?>')
def clean_htmlTags(text):
    return re.sub(pattern, '', text)

data['review'] = data['review'].apply(clean_htmlTags)
data['review'][0]
```

```python
def lower_case(text):
    return text.lower()

data['review'] = data['review'].apply(lower_case)
data['review'][0]
```

```python
def remove_splChar(text):
    return re.sub(r"[^\w\s']", ' ', text)
data['review'] = data['review'].apply(remove_splChar)
data['review'][0]
```

```python
import nltk
from nltk.corpus import stopwords
stopWords = set(stopwords.words('english'))
```

```python
def remove_stopwords(text):
    x = []
    for i in text.split():
        if i not in stopWords:
            x.append(i)
    return ' '.join(x)
    
data['review'] = data['review'].apply(remove_stopwords)
data['review'][0]
```

```python
from nltk.stem.porter import PorterStemmer
ps = PorterStemmer()

def stem(text):
    x = []
    for i in text.split():
        x.append(ps.stem(i))
    return ' '.join(x)

data['review'] = data['review'].apply(stem)
data['review'][0]
```

```python
data = data.iloc[:10000]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv = CountVectorizer(max_features = 5000) 
X = cv.fit_transform(data['review']).toarray()
```

```python
data.rename(columns={'sentiment': 'label'}, inplace=True)
data['label'] = data['label'].map({
    'positive': 1,
    'negative': 0
})
```

```python
y = data.iloc[:, -1]
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2)
```

```python
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
clf1 = GaussianNB()
clf2 = MultinomialNB()
clf3 = BernoulliNB()

clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)
clf3.fit(X_train, y_train)
```

```python
pred1 = clf1.predict(X_test)
pred2 = clf2.predict(X_test)
pred3 = clf3.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print("Gaussian", accuracy_score(y_test, pred1))
print("Multinomial", accuracy_score(y_test, pred2))
print("Bernoulli", accuracy_score(y_test, pred3))
```
